# Week 1 EDA — UCI Diabetes 130-US Hospitals

Target: `readmitted == "<30"` as 30-day readmission. Dataset path: `../data/diabetic_data.csv`.

```python
import pandas as pd
from pathlib import Path

csv_path = Path("../data/diabetic_data.csv")
df = pd.read_csv(csv_path)
print(df.shape)
print(df.columns.tolist()[:20], "...")
print(df["readmitted"].value_counts(normalize=True))
print("missing % (top 10)")
print((df == "?").mean().sort_values(ascending=False).head(10))
```
